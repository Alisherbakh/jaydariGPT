```python
text = 'Hello world this is a simple example to show how tokenization works in NLP and a sentence'   # 17 token   → ishlov beriladigan matn (gap)
print(text)
```

```python
tokens = text.split()             # Matnni bo'sh joylar bo'yicha so'zlarga bo'lamiz → so'zlar ro'yxati (tokenlar)
print(tokens)                     # → ['Hello', 'world', 'this', ..., 'sentence']  (17 ta)
```

```python
vocab = {}                        # Lug'at (vocabulary): so'z → raqam
current_id = 0                    # Keyingi yangi so'zga beriladigan raqam (0 dan boshlanadi)

for token in tokens:              # Har bir so'zni birma-bir olamiz
  if token not in vocab:          # Agar bu so'z lug'atda hali yo'q bo'lsa...
    vocab[token] = current_id     # ...unga hozirgi raqamni beramiz
    current_id += 1               # ...va keyingi so'z uchun raqamni 1 ga oshiramiz

print(vocab)
```

```python
encoded = [vocab[token] for token in tokens]   # Har bir so'zni lug'atdagi raqamiga almashtiramiz (encode)
print(encoded)
```

```python
reverse_vocab = {id: token for token, id in vocab.items()}   # Lug'atni teskari aylantiramiz: raqam → so'z  {0: 'Hello', 1: 'world', ...}
decoded = " ".join([reverse_vocab[id] for id in encoded])    # Har bir raqamni so'zga aylantirib, bo'sh joy bilan ulaymiz
print(decoded)                                               # → asl matn qaytdi: "Hello world this is ..."
```

```python
# Embedding vector => dense numerical array | Vector Database   → embedding — zich sonlar massivi; ular Vector Database'da saqlanadi
# 0 => [0.41, 0.22, -0.77, 0.56.........]                       → token ID 0 ('Hello') ning vektori
# 1 => [0.34, 0.72, 0.77, 0.82.........]                        → token ID 1 ('world') ning vektori
```

```python
import torch                      # PyTorch kutubxonasi
import torch.nn as nn             # Neyron tarmoq qatlamlari (Embedding shu yerda)

embedding_dim = 8                 # ⚠️ Har bir so'z vektorining uzunligi — 0 bo'lgani uchun natija bo'sh chiqdi
vocab_size = len(vocab)           # Lug'atdagi noyob so'zlar soni → 16

embedding_layer = nn.Embedding(vocab_size, embedding_dim)   # 16 ta so'z uchun jadval yaratamiz: har biriga embedding_dim uzunlikdagi tasodifiy vektor

input_ids = torch.tensor([encoded])           # Raqamlar ro'yxatini tensorga aylantiramiz; [ ] → batch o'lchami qo'shiladi → shape [1, 17]
embeddings = embedding_layer(input_ids)       # Har bir raqam o'rniga uning vektorini qo'yamiz → shape [1, 17, embedding_dim]

print(embeddings)                 # Vektorlarni chiqaramiz
print(embeddings.shape)           # Shaklini chiqaramiz
```

```python
from transformers import AutoTokenizer      # Hugging Face kutubxonasidan tayyor tokenizer yuklovchi klassni chaqiramiz

model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"     # Hugging Face'dagi model nomi: "muallif/model_nomi"
tokenizer = AutoTokenizer.from_pretrained(model_id) # Shu modelga mos tokenizerni internetdan yuklab olamiz
```

```python
print('Vocab size:', tokenizer.vocab_size)                  # Lug'atdagi tokenlar soni → 32000
print('Special tokens:', tokenizer.special_tokens_map)      # Maxsus tokenlar ro'yxati (gap boshi, oxiri, noma'lum, to'ldiruvchi)
```

```python
text = "Hello how are you doing?"         # Tokenize qilinadigan matn
print('text:', text)                      # Matnni chiqaramiz
```

```python
tokens = tokenizer.tokenize(text)         # Matnni tokenlarga bo'lamiz (hali raqam emas, matn bo'laklari)
print('tokens:', tokens)                  # → ['▁Hello', '▁how', '▁are', '▁you', '▁doing', '?']
```

```python
ids = tokenizer.convert_tokens_to_ids(tokens)   # Har bir tokenni lug'atdagi ID raqamiga almashtiramiz
print('token IDs:', ids)                  # → [15043, 920, 526, 366, 2599, 29973]
```

```python
encoded = tokenizer(text)                 # Hammasini bir qadamda: tokenize + ID + maxsus tokenlar + attention_mask
print('encoded:', encoded)                # → {'input_ids': [1, 15043, ...], 'attention_mask': [1, 1, ...]}
```

```python
print('Input IDs:', encoded['input_ids'])      # encoded lug'atidan faqat ID'lar ro'yxatini olamiz → [1, 15043, 920, ...]
```

```python
"Hi" #=> [1, 345, 2, 2, 2] // length 5                    # Kalta gap: oxiri pad tokeni (2) bilan to'ldirilib, 5 uzunlikka keltirilgan
"How are you?" #=> [1, 45, 657, 234, 23453] // length 5   # Uzunroq gap: to'ldirish kerak emas, o'zi 5 ta
# 'attention_mask': [1, 1, 0, 0, 0]                       # "Hi" uchun: 1 → haqiqiy token, 0 → padding (modelga "bularga qarama")
```

```python
decoded_text = tokenizer.decode(encoded["input_ids"])     # ID'lardan qaytadan matn yasaymiz
print('decoded_text:', decoded_text)                      # → "<s> Hello how are you doing?"  (<s> = gap boshi tokeni)
```

```python
texts = [                                                 # Bir nechta gapdan iborat ro'yxat (batch)
    "Hello, how are you?",                                # 1-gap (qisqa)
    "Today we are learning about tokenizers.",            # 2-gap
    "TinyLlama is a small but powerful language model."   # 3-gap (eng uzun)
]

batch = tokenizer(texts, padding=True, truncation=True, return_tensors="pt")   # 3 ta gapni birdaniga tokenize qilamiz
print(batch.keys())                                       # Natija ichidagi kalitlar → input_ids, attention_mask


print("Input IDs shape:", batch["input_ids"].shape)          # → torch.Size([3, N])  3 ta gap, N = eng uzun gapning token soni
print('Attention mask shape:', batch["attention_mask"].shape) # → torch.Size([3, N])  input_ids bilan bir xil shakl
```

```python
word = "headerandshoulder"   # sub word   → lug'atda yo'q, bo'sh joysiz qo'shib yozilgan "so'z"
print("word:", word)                             # So'zni chiqaramiz
print("Tokens:", tokenizer.tokenize(word))       # Tokenizer uni tanish bo'laklarga ajratadi → ['▁header', 'and', 'should', 'er']
print("IDs:", tokenizer.encode(word))            # Bo'laklarni ID'larga aylantiradi (boshida <s> = 1 bilan) → [1, 4839, 392, 9344, 261]
```
