```python
text = 'Hello world this is a simple example to show how tokenization works in NLP and a sentence'   # 17 token   → ishlov beriladigan matn (gap)
print(text)
```

```python
tokens = text.split()             # Matnni bo'sh joylar bo'yicha so'zlarga bo'lamiz → so'zlar ro'yxati (tokenlar)
print(tokens)                     # → ['Hello', 'world', 'this', ..., 'sentence']  (17 ta)
```

```python
vocab = {}                        # Lug'at (vocabulary): so'z → raqam
current_id = 0                    # Keyingi yangi so'zga beriladigan raqam (0 dan boshlanadi)

for token in tokens:              # Har bir so'zni birma-bir olamiz
  if token not in vocab:          # Agar bu so'z lug'atda hali yo'q bo'lsa...
    vocab[token] = current_id     # ...unga hozirgi raqamni beramiz
    current_id += 1               # ...va keyingi so'z uchun raqamni 1 ga oshiramiz

print(vocab)
```

```python
encoded = [vocab[token] for token in tokens]   # Har bir so'zni lug'atdagi raqamiga almashtiramiz (encode)
print(encoded)
```

```python
reverse_vocab = {id: token for token, id in vocab.items()}   # Lug'atni teskari aylantiramiz: raqam → so'z  {0: 'Hello', 1: 'world', ...}
decoded = " ".join([reverse_vocab[id] for id in encoded])    # Har bir raqamni so'zga aylantirib, bo'sh joy bilan ulaymiz
print(decoded)                                               # → asl matn qaytdi: "Hello world this is ..."
```

```python
# Embedding vector => dense numerical array | Vector Database   → embedding — zich sonlar massivi; ular Vector Database'da saqlanadi
# 0 => [0.41, 0.22, -0.77, 0.56.........]                       → token ID 0 ('Hello') ning vektori
# 1 => [0.34, 0.72, 0.77, 0.82.........]                        → token ID 1 ('world') ning vektori
```

```python
import torch                      # PyTorch kutubxonasi
import torch.nn as nn             # Neyron tarmoq qatlamlari (Embedding shu yerda)

embedding_dim = 8                 # ⚠️ Har bir so'z vektorining uzunligi — 0 bo'lgani uchun natija bo'sh chiqdi
vocab_size = len(vocab)           # Lug'atdagi noyob so'zlar soni → 16

embedding_layer = nn.Embedding(vocab_size, embedding_dim)   # 16 ta so'z uchun jadval yaratamiz: har biriga embedding_dim uzunlikdagi tasodifiy vektor

input_ids = torch.tensor([encoded])           # Raqamlar ro'yxatini tensorga aylantiramiz; [ ] → batch o'lchami qo'shiladi → shape [1, 17]
embeddings = embedding_layer(input_ids)       # Har bir raqam o'rniga uning vektorini qo'yamiz → shape [1, 17, embedding_dim]

print(embeddings)                 # Vektorlarni chiqaramiz
print(embeddings.shape)           # Shaklini chiqaramiz
```
